```python
#wap to count vowels and consonants in a string
name ='prashant'
data=['a','e','i','o','u']
vowel=0
cons=0

for i in name:
  if i in data:
    vowel+=1
  else:
    cons+=1

print(f"Number of vowels in name {name} is {vowel}")
print(f"Number of consonantsin name {name} is {cons}")
```

```python
#wap to remove duplicate characters in a string and only print unique chars

name = "guruden"
newname=''

for i in name:
  if i not in newname:
    newname+=i

print(name)
print(newname)
```
